```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# --- Set a clean, academic style ---
sns.set_style("white")
sns.set_context("paper") # Scales fonts and elements for a paper
plt.rcParams['figure.dpi'] = 150 # Sharper output

# --- 1. Mock Data Preparation ---

model_names_math500 = binary_df_math500_modified.index
zero_shot_mask = np.array([name.endswith("_zero_shot") for name in model_names_math500])
theta_lart_math500_zero = result_math500['theta_joint'].iloc[0][~gemma_27b_indices_bool][zero_shot_mask]
theta_irt_math500_zero = result_math500['theta_irt'].iloc[0][~gemma_27b_indices_bool][zero_shot_mask]

order_lart_math500 = np.argsort(-theta_lart_math500_zero)  # Descending order
order_irt_math500 = np.argsort(-theta_irt_math500_zero)    # Descending order
ranking_lart_math500 = model_names_math500[zero_shot_mask][order_lart_math500]
ranking_irt_math500 = model_names_math500[zero_shot_mask][order_irt_math500]


# Create a DataFrame for easier lookup of positions
df_lart = pd.DataFrame({'Model': ranking_lart_math500, 'LaRT_Rank': range(len(ranking_lart_math500))})
df_irt = pd.DataFrame({'Model': ranking_irt_math500, 'IRT_Rank': range(len(ranking_irt_math500))})

# Merge the two rankings based on model name
merged_rankings = pd.merge(df_lart, df_irt, on='Model', how='inner')

# Sort by LaRT rank for consistent plotting order on the left
merged_rankings = merged_rankings.sort_values(by='LaRT_Rank').reset_index(drop=True)

# === MODIFICATION: Filter for shifted models *before* plotting ===
shifted_rankings = merged_rankings[merged_rankings['LaRT_Rank'] != merged_rankings['IRT_Rank']].copy()
num_shifted_models = len(shifted_rankings)

# If no models shifted, print a message and show an empty plot
if num_shifted_models == 0:
    print("No models shifted in rank. Plot will not be generated.")
    fig, ax = plt.subplots(figsize=(10, 5))
    ax.text(0.5, 0.5, 'No models shifted in rank.', ha='center', va='center', fontsize=15)
    ax.set_xticks([])
    ax.set_yticks([])
    # Remove all spines
    for spine in ax.spines.values():
        spine.set_visible(False)
    plt.title('Comparison of Model Rankings (LaRT vs. IRT)', fontsize=16, pad=20, weight='bold')
    plt.tight_layout()
    plt.savefig('ranking_comparison.pdf')
    plt.show()

else:
    # --- Create new plot ranks for only shifted models ---
    
    # Create ordered lists of *only* the shifted models
    left_order = shifted_rankings.sort_values(by='LaRT_Rank')['Model'].tolist()
    right_order = shifted_rankings.sort_values(by='IRT_Rank')['Model'].tolist()

    # Create a mapping from model name to its new y-position (plot rank)
    # The y-position is its index in the sorted list of *shifted* models
    plot_y_left_map = {model: i for i, model in enumerate(left_order)}
    plot_y_right_map = {model: i for i, model in enumerate(right_order)}


    # --- 2. Plot Setup ---
    # === MODIFICATION: Adjust height to be tighter ===
    # This gives 0.45 inches per model, with a minimum height of 6 inches.
    plot_height = max(2, num_shifted_models * 0.8) 
    fig, ax = plt.subplots(figsize=(10, plot_height)) # Adjust figure size

    # Define colors
    colors = sns.color_palette('colorblind') # Define color palette
    line_color = colors[0]  # Use first color from palette
    text_color = 'black'

    # --- 3. Plotting Models (Left & Right) and Connecting Lines ---

    # Plot model names on the left (LaRT)
    for i, model in enumerate(left_order):
        ax.text(
            x=-0.02, y=i, s=model.replace('_zero_shot', ''), # Clean label, add padding
            ha='right', va='center', 
            fontsize=12,       # <-- FONT SIZE for left model names
            color=text_color, 
            weight='normal'    # <-- FONT WEIGHT for left model names
        )

    # Plot model names on the right (IRT)
    for i, model in enumerate(right_order):
        ax.text(
            x=1.02, y=i, s=model.replace('_zero_shot', ''), # Clean label, add padding
            ha='left', va='center', 
            fontsize=12,       # <-- FONT SIZE for right model names
            color=text_color, 
            weight='normal'    # <-- FONT WEIGHT for right model names
        )

    # Draw the connecting lines
    for model in plot_y_left_map.keys(): # Iterate through all shifted models
        y_left = plot_y_left_map[model]
        y_right = plot_y_right_map[model]
        
        ax.plot(
            [0, 1], # x-coordinates (0 for left, 1 for right)
            [y_left, y_right], # y-coordinates (new plot ranks)
            color=line_color, 
            linewidth=1.5, # Thinner line
            alpha=0.6,     # Slightly more transparent
            marker='o',    # Add markers
            markersize=4   # Smaller markers
        )


    # --- 4. Customizing the Plot ---

    # Set y-axis limits to match the number of *shifted* models
    ax.set_ylim(-1, num_shifted_models)

    # Invert y-axis so rank 0 (best) is at the top
    ax.invert_yaxis()

    # Set x-axis limits and hide default ticks/labels
    ax.set_xlim(-0.5, 1.5) # Widen x-lim for padded text
    ax.set_xticks([]) # Hide x-axis ticks
    ax.set_yticks([]) # Hide y-axis ticks (we're using text labels)

    # Add titles for the ranking mechanisms
    ax.text(
        0, -0.5, 'LaRT Ranking', 
        ha='right', va='bottom', 
        fontsize=14,   # <-- FONT SIZE for left title
        weight='normal'  # <-- FONT WEIGHT for left title
    )
    ax.text(
        1, -0.5, 'IRT Ranking', 
        ha='left', va='bottom', 
        fontsize=14,   # <-- FONT SIZE for right title
        weight='normal'  # <-- FONT WEIGHT for right title
    )

    # Remove spines (the box around the plot)
    ax.spines['left'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['top'].set_visible(False)
    ax.spines['bottom'].set_visible(False)

    # === MODIFICATION: Make axis lines fainter ===
    ax.axvline(x=0, color='lightgray', linestyle='--', linewidth=1)
    ax.axvline(x=1, color='lightgray', linestyle='--', linewidth=1)

    plt.title(
        'Comparison of Rankings for MATH500 for Zero-Shot Models', 
        fontsize=16,   # <-- FONT SIZE for main title
        pad=20,
        weight='normal'  # <-- FONT WEIGHT for main title (e.g., 'normal', 'bold')
    )
    plt.tight_layout()
    # Save as PDF for the paper
    plt.savefig('../figures/app_2/lart_irt_ranking_math500_0.pdf', bbox_inches='tight')
    plt.show()

```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# --- Set a clean, academic style ---
sns.set_style("white")
sns.set_context("paper") # Scales fonts and elements for a paper
plt.rcParams['figure.dpi'] = 150 # Sharper output
colors = sns.color_palette('bright') # Define color palette

# === Reusable Plotting Function ===
def create_slope_graph(ax, shifted_rankings, line_color, suffix_to_remove, title, fontsize=16):
    """
    Draws a slope graph for model ranking comparisons on a given matplotlib Axes.
    """
    num_shifted_models = len(shifted_rankings)
    
    # === MODIFICATION: Adjust horizontal spacing ===
    # To make the space WIDER: decrease x_left, increase x_right
    # To make the space SMALLER: increase x_left, decrease x_right
    x_left = 0.43
    x_right = 0.57
    text_pad = 0.02 # Padding for model labels
    # ============================================

    # If no models shifted, print a message and show an empty plot
    if num_shifted_models == 0:
        ax.text(0.5, 0.5, 'No models shifted in rank.', ha='center', va='center', fontsize=14)
        ax.set_xticks([])
        ax.set_yticks([])
        # Remove all spines
        for spine in ax.spines.values():
            spine.set_visible(False)
        ax.set_title(title, fontsize=16, pad=20, weight='normal')
        return

    # --- Create new plot ranks for only shifted models ---
    left_order = shifted_rankings.sort_values(by='LaRT_Rank')['Model'].tolist()
    right_order = shifted_rankings.sort_values(by='IRT_Rank')['Model'].tolist()

    plot_y_left_map = {model: i for i, model in enumerate(left_order)}
    plot_y_right_map = {model: i for i, model in enumerate(right_order)}

    # --- 3. Plotting Models (Left & Right) and Connecting Lines ---
    text_color = 'black'
    
    # Plot model names on the left (LaRT)
    for i, model in enumerate(left_order):
        ax.text(
            x=x_left - text_pad, y=i, s=model.replace(suffix_to_remove, ''), # Use x_left
            ha='right', va='center', 
            fontsize=13, color=text_color, weight='normal'
        )

    # Plot model names on the right (IRT)
    for i, model in enumerate(right_order):
        ax.text(
            x=x_right + text_pad, y=i, s=model.replace(suffix_to_remove, ''), # Use x_right
            ha='left', va='center', 
            fontsize=13, color=text_color, weight='normal'
        )

    # Draw the connecting lines
    for model in plot_y_left_map.keys():
        y_left = plot_y_left_map[model]
        y_right = plot_y_right_map[model]
        
        ax.plot(
            [x_left, x_right], [y_left, y_right], # Use x_left and x_right
            color=line_color, 
            linewidth=1.5, alpha=0.6,
            marker='o', markersize=4
        )

    # --- 4. Customizing the Plot ---
    ax.set_ylim(-1, num_shifted_models)
    ax.invert_yaxis()
    # Adjust xlim based on new coordinates and padding
    ax.set_xlim(x_left - 0.5, x_right + 0.5) 
    ax.set_xticks([])
    ax.set_yticks([])

    # Add titles for the ranking mechanisms
    ax.text(x_left, -0.5, 'LaRT Ranking', ha='right', va='bottom', fontsize=18, weight='normal')
    ax.text(x_right, -0.5, 'IRT Ranking', ha='left', va='bottom', fontsize=18, weight='normal')

    # Remove spines (the box around the plot)
    for spine in ax.spines.values():
        spine.set_visible(False)

    # Make axis lines fainter
    ax.axvline(x=x_left, color='lightgray', linestyle='--', linewidth=1) # Use x_left
    ax.axvline(x=x_right, color='lightgray', linestyle='--', linewidth=1) # Use x_right

    ax.set_title(title, fontsize=22, pad=20, weight='normal')

# === End of Function ===


# --- 1. Data Preparation (Both Plots) ---

# --- Plot 1: Zero-Shot Models ---
model_names_math500 = binary_df_math500_modified.index
zero_shot_mask = np.array([name.endswith("_zero_shot") for name in model_names_math500])
theta_lart_math500_zero = result_math500['theta_joint'].iloc[0][~gemma_27b_indices_bool][zero_shot_mask]
theta_irt_math500_zero = result_math500['theta_irt'].iloc[0][~gemma_27b_indices_bool][zero_shot_mask]

order_lart_math500_zero = np.argsort(-theta_lart_math500_zero)
order_irt_math500_zero = np.argsort(-theta_irt_math500_zero)
ranking_lart_math500_zero = model_names_math500[zero_shot_mask][order_lart_math500_zero]
ranking_irt_math500_zero = model_names_math500[zero_shot_mask][order_irt_math500_zero]

df_lart_zero = pd.DataFrame({'Model': ranking_lart_math500_zero, 'LaRT_Rank': range(len(ranking_lart_math500_zero))})
df_irt_zero = pd.DataFrame({'Model': ranking_irt_math500_zero, 'IRT_Rank': range(len(ranking_irt_math500_zero))})
merged_rankings_zero = pd.merge(df_lart_zero, df_irt_zero, on='Model', how='inner')
shifted_rankings_zero = merged_rankings_zero[merged_rankings_zero['LaRT_Rank'] != merged_rankings_zero['IRT_Rank']].copy()
num_shifted_zero = len(shifted_rankings_zero)

# --- Plot 2: One-Shot Models ---
one_shot_mask = np.array([name.endswith("_one_shot") for name in model_names_math500])
theta_lart_math500_one = result_math500['theta_joint'].iloc[0][~gemma_27b_indices_bool][one_shot_mask]
theta_irt_math500_one = result_math500['theta_irt'].iloc[0][~gemma_27b_indices_bool][one_shot_mask]

order_lart_math500_one = np.argsort(-theta_lart_math500_one)
order_irt_math500_one = np.argsort(-theta_irt_math500_one)
ranking_lart_math500_one = model_names_math500[one_shot_mask][order_lart_math500_one]
ranking_irt_math500_one = model_names_math500[one_shot_mask][order_irt_math500_one]

df_lart_one = pd.DataFrame({'Model': ranking_lart_math500_one, 'LaRT_Rank': range(len(ranking_lart_math500_one))})
df_irt_one = pd.DataFrame({'Model': ranking_irt_math500_one, 'IRT_Rank': range(len(ranking_irt_math500_one))})
merged_rankings_one = pd.merge(df_lart_one, df_irt_one, on='Model', how='inner')
shifted_rankings_one = merged_rankings_one[merged_rankings_one['LaRT_Rank'] != merged_rankings_one['IRT_Rank']].copy()
num_shifted_one = len(shifted_rankings_one)


# --- 2. Create the Combined Plot ---

# Calculate plot height based on the *max* number of models to show
# This keeps both plots the same height.
# You can adjust the `0.35` multiplier to change spacing
max_shifted_models = max(num_shifted_zero, num_shifted_one, 1) # Use 1 to avoid 0
plot_height = max(3, max_shifted_models * 0.35) 

# Create a 1-row, 2-column figure.
# Figure width is 20 (10 per plot)
fig, axes = plt.subplots(1, 2, figsize=(14, plot_height))

# --- 3. Draw Each Plot ---

# Draw Plot 1 (Zero-Shot) on the left axis (axes[0])
create_slope_graph(
    ax=axes[0], 
    shifted_rankings=shifted_rankings_zero, 
    line_color=colors[0], 
    suffix_to_remove='_zero_shot',
    title='Zero-Shot Models'
)

# Draw Plot 2 (One-Shot) on the right axis (axes[1])
create_slope_graph(
    ax=axes[1], 
    shifted_rankings=shifted_rankings_one, 
    line_color=colors[1], # Use a different color
    suffix_to_remove='_one_shot',
    title='One-Shot Models'
)

# --- 4. Finalize and Save ---
fig.suptitle(
    'Comparison of Model Rankings for MATH500', 
    fontsize=24, 
    weight='bold'
)

# Adjust layout to prevent title overlap
plt.tight_layout(rect=[0, 0, 1, 0.95]) # Make room for suptitle

# Save the combined figure as a single PDF
plt.savefig('../figures/app_2/lart_irt_ranking_math500_combined.pdf', bbox_inches='tight')
plt.show()

```

```python
def kendalls_w(rankings):
    """
    Calculate Kendall's W (Coefficient of Concordance).

    Args:
        rankings (np.ndarray): A 2D NumPy array where rows are the items 
                               being ranked and columns are the rankers (judges).

    Returns:
        float: The Kendall's W statistic, ranging from 0 (no agreement) 
               to 1 (perfect agreement).
    """
    num_items, num_rankers = rankings.shape
    rank_sums = np.sum(rankings, axis=1)

    mean_rank_sum = np.mean(rank_sums)
    S = np.sum((rank_sums - mean_rank_sum)**2)

    max_S = (num_rankers**2 * (num_items**3 - num_items)) / 12

    W = S / max_S
    
    return W
```

```python
# Calculate ranks for each subarray in descending order (largest value gets rank 1)
theta_irt_ranks = []
for theta_list in estimated_params_irt_flat_parallel["theta_irt"]:
    theta_array = np.array(theta_list)
    # argsort returns indices that would sort the array, so we reverse for descending
    ranks = np.argsort(-theta_array) + 1  # rank starts at 1
    theta_irt_ranks.append(ranks)

theta_irt_ranks = np.array(theta_irt_ranks)
theta_irt_ranks = theta_irt_ranks.T
```

```python
# Calculate ranks for each subarray in descending order (largest value gets rank 1)
theta_joint_ranks = []
for theta_list in estimated_params_joint_flat_parallel["theta_joint"]:
    theta_array = np.array(theta_list)
    # argsort returns indices that would sort the array, so we reverse for descending
    ranks = np.argsort(-theta_array) + 1  # rank starts at 1
    theta_joint_ranks.append(ranks)

theta_joint_ranks = np.array(theta_joint_ranks)
theta_joint_ranks = theta_joint_ranks.T
```

```python
w_IRT = kendalls_w(theta_irt_ranks)
w_Joint = kendalls_w(theta_joint_ranks)

print(f"Kendall's W for IRT ranks: {w_IRT:.4f}")
print(f"Kendall's W for Joint ranks: {w_Joint:.4f}")
```

```python
from scipy.stats import norm

def fisher_info(a, b, theta):
    '''Compute Item-wise log Fisher Information for Normal-ogive model for each test taker (theta 1-dim)'''
    args = a[np.newaxis,:] * theta[:,np.newaxis] + b[np.newaxis,:]

    log_a = np.log(np.maximum(a, 1e-12))
    log_num = 2 * norm.logpdf(args) + 2 * log_a[np.newaxis,:]
    log_denom = norm.logcdf(args) + norm.logcdf(-args)
    log_fisher_info = log_num - log_denom
    fisher_info = np.exp(log_fisher_info)

    return fisher_info

fisher_info_lart = fisher_info(
    results_math500['a_joint'].iloc[0],
    results_math500['b_joint'].iloc[0],
    results_math500['theta_joint'].iloc[0]
)
fisher_info_irt = fisher_info(
    results_math500['a_irt'].iloc[0],
    results_math500['b_irt'].iloc[0],
    results_math500['theta_irt'].iloc[0]
)
```

```python
def hodges_lehmann(vec):
    x = np.asarray(vec)
    walsh_sums = np.add.outer(x, x) 
    walsh_averages = walsh_sums[np.triu_indices_from(walsh_sums)].flatten() / 2
    return np.median(walsh_averages)
```

```python
def fisher_info_theta_lart(a, b, theta, Sigma):
    val = theta[:, np.newaxis] * a[np.newaxis, :] + b[np.newaxis, :]
    log_phi_2 = 2 * norm.logpdf(val)
    log_Phi_p = norm.logcdf(val)
    log_Phi_m = norm.logcdf(-val)
    mat_common = np.exp(np.nan_to_num(log_phi_2 - log_Phi_p - log_Phi_m, neginf=0.0)) * (a ** 2)[np.newaxis, :]
    # mat_common = (np.exp(np.nan_to_num(log_phi_2 - log_Phi_p, neginf=0.0)) + \
                #    np.exp(np.nan_to_num(log_phi_2 - log_Phi_m, neginf=0.0))) * (a ** 2)[np.newaxis, :]
    fisher_info_1 = np.sum(mat_common, axis=1)

    Sigma_inv = np.linalg.inv(Sigma)
    fisher_info_2 = Sigma_inv[0, 0]
    return fisher_info_1 + fisher_info_2
```

```python
fisher_info_mat = np.zeros((len(theta_est), 500))
for j in range(500):
    b_j = b_est[j].reshape(-1, 1)
    a_j = a_est[j].reshape(-1, 1)
    fisher_info_mat[:,j] = fisher_info_theta_lart(a_j, b_j, theta_est, Sigma_est)
```

```python
fisher_info_per_question = np.mean(fisher_info_mat, axis=0)
```

```python
problem_type_unique = problem_types['problem_type'].unique()
## select the indxes with the largest fisher information for each problem type, separately and save the indices for each problem type
selected_indices = []
selected_indices_map = {}

print("--- Selecting Top 25 Questions per Type ---")

for p_type in problem_type_unique:
    # Get all indices belonging to this problem type
    type_indices = problem_types[problem_types['problem_type'] == p_type].index.values
    
    # Get the Fisher Info scores for these specific questions
    scores = fisher_info_per_question[type_indices]
    
    # Combine into a temporary DataFrame to sort easily
    temp_df = pd.DataFrame({'original_index': type_indices, 'fisher_info': scores})
    
    # Select top 25 (or all if less than 25 exist)
    top_25 = temp_df.nlargest(25, 'fisher_info')
    
    # Store the original indices
    selected_indices_map[p_type] = top_25['original_index'].values
```
