```python
from matplotlib.lines import Line2D
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# 1. Prepare the data (Melting the dataframes for theta, a, and b)
df_theta = pd.DataFrame({
    "J": sim_varyJ["J"].tolist() * 2,
    "Method": ["LaRT"] * len(sim_varyJ) + ["IRT"] * len(sim_varyJ),
    "RMSE": sim_varyJ["rmse_theta"].tolist() + sim_varyJ["rmse_theta_irt"].tolist()
})

df_a = pd.DataFrame({
    "J": sim_varyJ["J"].tolist() * 2,
    "Method": ["LaRT"] * len(sim_varyJ) + ["IRT"] * len(sim_varyJ),
    "RMSE": sim_varyJ["rmse_a"].tolist() + sim_varyJ["rmse_a_irt"].tolist()
})

df_b = pd.DataFrame({
    "J": sim_varyJ["J"].tolist() * 2,
    "Method": ["LaRT"] * len(sim_varyJ) + ["IRT"] * len(sim_varyJ),
    "RMSE": sim_varyJ["rmse_b"].tolist() + sim_varyJ["rmse_b_irt"].tolist()
})

colors = sns.color_palette("Set1")

# 2. Create the Figure and Subplots
fig, axes = plt.subplots(1, 3, figsize=(16, 5.5), sharex=True)

# Theta RMSE Plot
sns.boxplot(
    ax=axes[0],
    data=df_theta,
    x="J",
    y="RMSE",
    hue="Method",
    palette={"LaRT": colors[1], "IRT": colors[0]},
    hue_order=["LaRT", "IRT"]
)
axes[0].set_xlabel("J", fontsize=14)
axes[0].set_ylabel("RMSE", fontsize=14)
axes[0].set_title(r"$\theta$", fontsize=16)
axes[0].tick_params(labelsize=12)
axes[0].get_legend().remove() 

# a RMSE Plot
sns.boxplot(
    ax=axes[1],
    data=df_a,
    x="J",
    y="RMSE",
    hue="Method",
    palette={"LaRT": colors[1], "IRT": colors[0]},
    hue_order=["LaRT", "IRT"]
)
axes[1].set_xlabel("J", fontsize=14)
axes[1].set_ylabel("", fontsize=14)
axes[1].set_title("a", fontsize=16)
axes[1].tick_params(labelsize=12)
axes[1].get_legend().remove()

# b RMSE Plot
sns.boxplot(
    ax=axes[2],
    data=df_b,
    x="J",
    y="RMSE",
    hue="Method",
    palette={"LaRT": colors[1], "IRT": colors[0]},
    hue_order=["LaRT", "IRT"]
)
axes[2].set_xlabel("J", fontsize=14)
axes[2].set_ylabel("", fontsize=14)
axes[2].set_title("b", fontsize=16)
axes[2].tick_params(labelsize=12)


# 3. Modify and share the Legend
# Get handles and labels from the last plot before we remove it
handles, labels = axes[2].get_legend_handles_labels()
axes[2].get_legend().remove()

# Create a dummy handle (an invisible line) for the "Method" text
dummy_handle = Line2D([0], [0], marker='none', color='none', label='Method')

# Create new lists for handles and labels
new_handles = [dummy_handle, handles[0], handles[1]]
new_labels = ['Method', labels[0], labels[1]]

# Add a single, figure-level legend at the bottom
fig.legend(
    new_handles, 
    new_labels, 
    loc='lower center', # Position at the bottom center
    ncol=3,             # Arrange all 3 items in a single row
    fontsize=14,
    frameon=False       # Remove the box
)

# 4. Title and Layout Modification
fig.suptitle(r"Comparison between LaRT and IRT with Varying J", fontsize=20)
sns.despine()

# Adjust layout to make room for the bottom legend and top suptitle
plt.tight_layout(rect=[0, 0.1, 1, 0.95]) 

# Show or save the plot
plt.show()
plt.savefig("./Analysis/Fig/sim_rev/com_acc_varyJ.pdf") # Uncomment to save
```

```python
from matplotlib.lines import Line2D
from matplotlib.patches import PathPatch
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# 1. Prepare the data (Melting the dataframes for theta, a, and b)
# Assuming sim_varyJ is already defined
df_theta = pd.DataFrame({
    "J": sim_varyJ["J"].tolist() * 2,
    "Method": ["LaRT"] * len(sim_varyJ) + ["IRT"] * len(sim_varyJ),
    "RMSE": sim_varyJ["rmse_theta"].tolist() + sim_varyJ["rmse_theta_irt"].tolist()
})

df_a = pd.DataFrame({
    "J": sim_varyJ["J"].tolist() * 2,
    "Method": ["LaRT"] * len(sim_varyJ) + ["IRT"] * len(sim_varyJ),
    "RMSE": sim_varyJ["rmse_a"].tolist() + sim_varyJ["rmse_a_irt"].tolist()
})

df_b = pd.DataFrame({
    "J": sim_varyJ["J"].tolist() * 2,
    "Method": ["LaRT"] * len(sim_varyJ) + ["IRT"] * len(sim_varyJ),
    "RMSE": sim_varyJ["rmse_b"].tolist() + sim_varyJ["rmse_b_irt"].tolist()
})

# Revert to original Set1 colors
colors = sns.color_palette("Set1")
original_palette = {"LaRT": colors[1], "IRT": colors[0]}

# 2. Create the Figure and Subplots
fig, axes = plt.subplots(1, 3, figsize=(16, 5.5), sharex=True)

# Define a helper function to add hatches to a specific axis
def add_hatches(ax):
    # Extract only the box patches
    boxes = [patch for patch in ax.patches if type(patch).__name__ in ['PathPatch', 'Rectangle']]
    
    # Calculate the number of J categories
    num_categories = len(sim_varyJ["J"].unique())
    
    for i, box in enumerate(boxes):
        if i >= num_categories: 
            # These correspond to the second hue category (IRT)
            box.set_hatch('//')         # Add diagonal lines
            box.set_edgecolor('black')  # Ensure the hatch/edge is clearly visible

# Theta RMSE Plot
sns.boxplot(
    ax=axes[0],
    data=df_theta,
    x="J",
    y="RMSE",
    hue="Method",
    palette=original_palette,
    hue_order=["LaRT", "IRT"],
    linecolor='black'
)
axes[0].set_xlabel("J", fontsize=14)
axes[0].set_ylabel("RMSE", fontsize=14)
axes[0].set_title(r"$\theta$", fontsize=16)
axes[0].tick_params(labelsize=12)
axes[0].get_legend().remove() 
add_hatches(axes[0]) # Apply hatching

# a RMSE Plot
sns.boxplot(
    ax=axes[1],
    data=df_a,
    x="J",
    y="RMSE",
    hue="Method",
    palette=original_palette,
    hue_order=["LaRT", "IRT"],
    linecolor='black'
)
axes[1].set_xlabel("J", fontsize=14)
axes[1].set_ylabel("", fontsize=14)
axes[1].set_title("a", fontsize=16)
axes[1].tick_params(labelsize=12)
axes[1].get_legend().remove()
add_hatches(axes[1]) # Apply hatching

# b RMSE Plot
sns.boxplot(
    ax=axes[2],
    data=df_b,
    x="J",
    y="RMSE",
    hue="Method",
    palette=original_palette,
    hue_order=["LaRT", "IRT"],
    linecolor='black'
)
axes[2].set_xlabel("J", fontsize=14)
axes[2].set_ylabel("", fontsize=14)
axes[2].set_title("b", fontsize=16)
axes[2].tick_params(labelsize=12)
add_hatches(axes[2]) # Apply hatching


# 3. Modify and share the Legend
axes[2].get_legend().remove()

# Create custom handles to match the new styles and colors exactly
dummy_handle = Line2D([0], [0], marker='none', color='none', label='Method')

# Handle for LaRT (colors[1] = Blue, no hatch)
lart_handle = plt.Rectangle((0, 0), 1, 1, facecolor=colors[1], edgecolor='black')

# Handle for IRT (colors[0] = Red, diagonal hatch)
irt_handle = plt.Rectangle((0, 0), 1, 1, facecolor=colors[0], edgecolor='black', hatch='//')

new_handles = [dummy_handle, lart_handle, irt_handle]
new_labels = ['Method', 'LaRT', 'IRT']

# Add a single, figure-level legend at the bottom
fig.legend(
    new_handles, 
    new_labels, 
    loc='lower center', 
    ncol=3,             
    fontsize=14,
    frameon=False       
)

# 4. Title and Layout Modification
fig.suptitle(r"Comparison between LaRT and IRT with Varying J", fontsize=20)
sns.despine()

# Adjust layout to make room for the bottom legend and top suptitle
plt.tight_layout(rect=[0, 0.1, 1, 0.95]) 

# Show or save the plot
plt.savefig("./Analysis/Fig/sim_rev/com_acc_varyJ.pdf")
plt.show()
 # Uncomment to save
```

```python
from matplotlib.lines import Line2D
from matplotlib.patches import PathPatch
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# 1. (Plot 1 data manipulation is commented out in your code, keeping it unchanged)
df_plot1 = sim_mis_probit_poisson.melt(
    id_vars=["N"], 
    value_vars=["spearman_theta", "spearman_a", "spearman_b"],
    var_name="Parameter",
    value_name="Spearman Correlation"
)
df_plot1["Parameter"] = df_plot1["Parameter"].replace({
    "spearman_theta": r"$\theta$", 
    "spearman_a": "a", 
    "spearman_b": "b"
})

# 2. Melt the dataframe for Plot 2 (omega, phi, tau)
df_plot2 = sim_mis_probit_poisson.melt(
    id_vars=["N"], 
    value_vars=["spearman_omega", "spearman_phi", "spearman_tau"],
    var_name="Parameter", 
    value_name="Spearman Correlation"
)
df_plot2["Parameter"] = df_plot2["Parameter"].replace({
    "spearman_omega": r"$\omega$", 
    "spearman_phi": r"$\phi$", 
    "spearman_tau": r"$\tau$"
})

# Define a consistent color palette
colors = sns.color_palette("Set1")
parameter_order = [r"$\omega$", r"$\phi$", r"$\tau$"]

# 3. Create the figure and axes (1 row, 2 columns as in your code)
fig, axes = plt.subplots(1, 2, figsize=(12, 5.5))

# --- Helper function for adding hatches to the parameter boxplot ---
def add_hatches_to_hue(ax, num_categories):
    boxes = [patch for patch in ax.patches if type(patch).__name__ in ['PathPatch', 'Rectangle']]
    
    # We have 3 hues: omega (no hatch), phi (//), tau (xx)
    hatches = ['', '//', 'xx']
    
    for i, box in enumerate(boxes):
        # Determine which hue group the box belongs to
        hue_idx = i // num_categories 
        
        if hue_idx < len(hatches) and hatches[hue_idx]:
            box.set_hatch(hatches[hue_idx])
            
        box.set_edgecolor('black') # Apply black borders for journal formatting

# --- Helper function for single plot edgecolors ---
def format_single_boxplot(ax):
    boxes = [patch for patch in ax.patches if type(patch).__name__ in ['PathPatch', 'Rectangle']]
    for box in boxes:
        box.set_edgecolor('black')


# --- Plot 2: omega, phi, tau ---
sns.boxplot(
    ax=axes[0],
    data=df_plot2,
    x="N",
    y="Spearman Correlation",
    hue="Parameter",
    palette=colors[:3], # Use the first 3 colors of Set1
    hue_order=parameter_order,
    linecolor='black'
)
axes[0].set_title("Latency Parameters", fontsize=16)
axes[0].set_xlabel("N", fontsize=14)
axes[0].set_ylabel("", fontsize=14) 
axes[0].tick_params(labelsize=12)

# Apply hatches to Plot 2 (calculate number of unique N's)
num_N_categories = len(sim_mis_probit_poisson["N"].unique())
add_hatches_to_hue(axes[0], num_N_categories)

# Custom Legend for Plot 2
axes[0].get_legend().remove()
dummy_handle = Line2D([0], [0], marker='none', color='none', label='Parameter')
omega_handle = plt.Rectangle((0, 0), 1, 1, facecolor=colors[0], edgecolor='black')
phi_handle   = plt.Rectangle((0, 0), 1, 1, facecolor=colors[1], edgecolor='black', hatch='//')
tau_handle   = plt.Rectangle((0, 0), 1, 1, facecolor=colors[2], edgecolor='black', hatch='xx')

axes[0].legend(
    [dummy_handle, omega_handle, phi_handle, tau_handle],
    ['Parameter', r'$\omega$', r'$\phi$', r'$\tau$'],
    loc='lower right',
    fontsize=12,
    frameon=False # Optional: remove frame for cleaner look
)

# --- Plot 3: rho (MAE) ---
sns.boxplot(
    ax=axes[1],
    data=sim_mis_probit_poisson,
    x="N",
    y="mae_rho",
    color=colors[1],
    linecolor='black'
)
axes[1].set_title(r"Correlation $\rho$", fontsize=16)
axes[1].set_xlabel("N", fontsize=14)
axes[1].set_ylabel("MAE", fontsize=14)
axes[1].tick_params(labelsize=12)
format_single_boxplot(axes[1]) # Apply black edges

# 4. Final Layout Adjustments
sns.despine()
plt.tight_layout()

plt.savefig("./Analysis/Fig/sim_rev/misspec_probit_poisson.pdf") # Uncomment to save
plt.show()
```

```python
from matplotlib.lines import Line2D
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# 1. (Plot 1 data manipulation is commented out in your code, keeping it unchanged)
df_plot1 = sim_mis_logistic_poisson_lin.melt(
    id_vars=["N"], 
    value_vars=["spearman_theta", "spearman_a", "spearman_b"],
    var_name="Parameter",
    value_name="Spearman Correlation"
)
df_plot1["Parameter"] = df_plot1["Parameter"].replace({
    "spearman_theta": r"$\theta$", 
    "spearman_a": "a", 
    "spearman_b": "b"
})

# 2. Melt the dataframe for Plot 2 (omega, phi, tau)
df_plot2 = sim_mis_logistic_poisson_lin.melt(
    id_vars=["N"], 
    value_vars=["spearman_omega", "spearman_phi", "spearman_tau"],
    var_name="Parameter", 
    value_name="Spearman Correlation"
)
df_plot2["Parameter"] = df_plot2["Parameter"].replace({
    "spearman_omega": r"$\omega$", 
    "spearman_phi": r"$\phi$", 
    "spearman_tau": r"$\tau$"
})

# Define a consistent color palette
colors = sns.color_palette("Set1")
parameter_order = [r"$\omega$", r"$\phi$", r"$\tau$"]

# 3. Create the figure and axes (1 row, 2 columns as in your code)
fig, axes = plt.subplots(1, 2, figsize=(12, 5.5))

# --- Helper function for adding hatches to the parameter boxplot ---
def add_hatches_to_hue(ax, num_categories):
    boxes = [patch for patch in ax.patches if type(patch).__name__ in ['PathPatch', 'Rectangle']]
    
    # We have 3 hues: omega (no hatch), phi (//), tau (xx)
    hatches = ['', '//', 'xx']
    
    for i, box in enumerate(boxes):
        # Determine which hue group the box belongs to
        hue_idx = i // num_categories 
        
        if hue_idx < len(hatches) and hatches[hue_idx]:
            box.set_hatch(hatches[hue_idx])
            
        box.set_edgecolor('black') # Apply black borders for journal formatting

# --- Helper function for single plot edgecolors ---
def format_single_boxplot(ax):
    boxes = [patch for patch in ax.patches if type(patch).__name__ in ['PathPatch', 'Rectangle']]
    for box in boxes:
        box.set_edgecolor('black')


# --- Plot 2: omega, phi, tau ---
sns.boxplot(
    ax=axes[0],
    data=df_plot2,
    x="N",
    y="Spearman Correlation",
    hue="Parameter",
    palette=colors[:3], # Use the first 3 colors of Set1
    hue_order=parameter_order,
    linecolor='black'
)
axes[0].set_title("Latency Parameters", fontsize=16)
axes[0].set_xlabel("N", fontsize=14)
axes[0].set_ylabel("", fontsize=14) 
axes[0].tick_params(labelsize=12)

# Apply hatches to Plot 2 (calculate number of unique N's)
num_N_categories = len(sim_mis_logistic_poisson_lin["N"].unique())
add_hatches_to_hue(axes[0], num_N_categories)

# Custom Legend for Plot 2
axes[0].get_legend().remove()
dummy_handle = Line2D([0], [0], marker='none', color='none', label='Parameter')
omega_handle = plt.Rectangle((0, 0), 1, 1, facecolor=colors[0], edgecolor='black')
phi_handle   = plt.Rectangle((0, 0), 1, 1, facecolor=colors[1], edgecolor='black', hatch='//')
tau_handle   = plt.Rectangle((0, 0), 1, 1, facecolor=colors[2], edgecolor='black', hatch='xx')

axes[0].legend(
    [dummy_handle, omega_handle, phi_handle, tau_handle],
    ['Parameter', r'$\omega$', r'$\phi$', r'$\tau$'],
    loc='lower right',
    fontsize=12,
    frameon=False # Optional: remove frame for cleaner look
)

# --- Plot 3: rho (MAE) ---
sns.boxplot(
    ax=axes[1],
    data=sim_mis_logistic_poisson_lin,
    x="N",
    y="mae_rho",
    color=colors[1],
    linecolor='black'
)
axes[1].set_title(r"Correlation $\rho$", fontsize=16)
axes[1].set_xlabel("N", fontsize=14)
axes[1].set_ylabel("MAE", fontsize=14)
axes[1].tick_params(labelsize=12)
format_single_boxplot(axes[1]) # Apply black edges

# 4. Final Layout Adjustments
sns.despine()
plt.tight_layout()

plt.savefig("./Analysis/Fig/sim_rev/misspec_logistic_poisson_latent_lin.pdf") # Uncomment to save
plt.show()
```
